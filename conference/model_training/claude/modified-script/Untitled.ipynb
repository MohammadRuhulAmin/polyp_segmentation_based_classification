```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(10, 5), dpi=300)
ax.set_xlim(0, 10)
ax.set_ylim(0, 5)
ax.axis('off')

# Box Drawing Helper
def draw_box(x, y, w, h, text, color):
    rect = patches.FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.2", facecolor=color, edgecolor="black", linewidth=1.5)
    ax.add_patch(rect)
    ax.text(x + w/2, y + h/2, text, ha='center', va='center', fontsize=10, fontweight='bold')

# Draw System Components
draw_box(0.5, 2, 2, 1, "Large Intestine\n(Colon Target Region)", "#FFE4C4")
draw_box(3.5, 2, 2, 1, "Colonoscope\n(Flexible Tube & Camera)", "#E6E6FA")
draw_box(6.5, 2, 2.5, 1, "Monitor Display\n(Real-time Endoscopic View)", "#D1E8E2")

# Annotations & Arrows
ax.annotate('', xy=(3.5, 2.5), xytext=(2.5, 2.5), arrowprops=dict(arrowstyle="->", lw=2, color='black'))
ax.annotate('', xy=(6.5, 2.5), xytext=(5.5, 2.5), arrowprops=dict(arrowstyle="->", lw=2, color='black'))

ax.text(3.0, 2.7, "Insertion", fontsize=9, ha='center', fontstyle='italic')
ax.text(6.0, 2.7, "Video Feed", fontsize=9, ha='center', fontstyle='italic')

plt.title("Schematic Diagram of Colonoscopy System Setup", fontsize=12, fontweight='bold', pad=20)
plt.tight_layout()

# Save for IEEE Paper
plt.savefig("colonoscopy_system_setup.png", dpi=300, bbox_inches='tight')
plt.savefig("colonoscopy_system_setup.pdf", bbox_inches='tight')
plt.show()
```

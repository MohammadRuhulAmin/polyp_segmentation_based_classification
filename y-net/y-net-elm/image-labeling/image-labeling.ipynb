```python
from PIL import Image, ImageDraw, ImageFont

def label_endoscopic_image(input_image_path, output_image_path, top_label_text, bottom_label_text, font_path='/path/to/your/arial.ttf'):
    """
    Adds top and bottom labels on solid bars to an endoscopic image.
    
    Args:
        input_image_path: Path to the source endoscopic image (without labels).
        output_image_path: Path to save the labeled image.
        top_label_text: Text for the top label.
        bottom_label_text: Text for the bottom label.
        font_path: Path to a TrueType font file (e.g., Arial).
    """
    try:
        # Load the source endoscopic image
        core_image = Image.open(input_image_path).convert('RGB')
        core_width, core_height = core_image.size

        # Define bar heights (as percentages of core image height, adjust as needed)
        # These are rough estimates from your example image.
        top_bar_height = int(core_height * 0.10) # 10% for the top
        bottom_bar_height = int(core_height * 0.20) # 20% for the bottom
        
        # Calculate new total dimensions
        final_width = core_width
        final_height = core_height + top_bar_height + bottom_bar_height

        # Define colors (from the example)
        bar_color = "#1A1A1A"  # Very dark grey
        text_color = "white"

        # Create a new, blank image with the dark grey background color
        labeled_image = Image.new('RGB', (final_width, final_height), color=bar_color)

        # Paste the core endoscopic image in the center
        labeled_image.paste(core_image, (0, top_bar_height))

        # --- Draw the Labels ---
        draw = ImageDraw.Draw(labeled_image)

        # Try to load the specified font with appropriate sizes
        # We need two different sizes for the top and bottom.
        try:
            # Scaled sizes based on image size. Adjust multipliers as needed.
            top_font_size = int(top_bar_height * 0.70)
            bottom_font_size = int(bottom_font_size = int(bottom_bar_height * 0.40)) 
            
            top_font = ImageFont.truetype(font_path, top_font_size)
            bottom_font = ImageFont.truetype(font_path, bottom_font_size)
        except IOError:
            print(f"Font file '{font_path}' not found. Using default font.")
            top_font = ImageFont.load_default()
            bottom_font = ImageFont.load_default()

        # Center and draw the Top Label
        top_bbox = draw.textbbox((0, 0), top_label_text, font=top_font)
        top_text_w = top_bbox[2] - top_bbox[0]
        top_text_x = (final_width - top_text_w) // 2
        top_text_y = (top_bar_height - top_bbox[3]) // 2  # Middle of top bar
        draw.text((top_text_x, top_text_y), top_label_text, fill=text_color, font=top_font)

        # Center and draw the Bottom Label
        bottom_bbox = draw.textbbox((0, 0), bottom_label_text, font=bottom_font)
        bottom_text_w = bottom_bbox[2] - bottom_bbox[0]
        bottom_text_x = (final_width - bottom_text_w) // 2
        bottom_text_y = (core_height + top_bar_height) + (bottom_bar_height - bottom_bbox[3]) // 2 # Middle of bottom bar
        draw.text((bottom_text_x, bottom_text_y), bottom_label_text, fill=text_color, font=bottom_font)

        # Save the result
        labeled_image.save(output_image_path)
        print(f"Labeled image saved to: {output_image_path}")

    except Exception as e:
        print(f"An error occurred: {e}")

# ==========================================
# Example Usage
# ==========================================
# 1. Update the image path and other details.
# 2. IMPORTANT: You must provide a valid font path (e.g., Arial on Windows, or from a package).
# On Linux/macOS, common font paths include: '/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf' or similar.

# Replace with the path to your source (non-labeled) image.
# For example, if you had a 'raw_endoscopy.png' image:
input_path = 'raw_endoscopy.png' 
# Make sure this is different from input_path to avoid overwriting.
output_path = 'labeled_result.png' 

top_text = "ETIS-LaribPolypDB"
bottom_text = "(a)"

# Replace with the path to a .ttf font file on your system.
font_path = '/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf' # Example Linux path
# font_path = 'C:\\Windows\\Fonts\\arial.ttf' # Example Windows path

label_endoscopic_image(input_path, output_path, top_text, bottom_text, font_path)
if __name__ == "__main__":
    input_file = "WhatsApp Image 2026-08-17 at 4.52.01 PM.jpeg"      # Replace with your actual file path
    output_file = "labeled_image.png"    # Replace with your desired output path
    
    add_journal_labels(
        input_path=input_file, 
        output_path=output_file, 
        header_text="ETIS", 
        footer_text="(a)"
    )
```
